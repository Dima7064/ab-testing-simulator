```python
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
%matplotlib inline
```

```python
df = pd.read_csv("../data/multiple_testing.csv")

control = df[df["group"] == "control"]
treatment = df[df["group"] == "treatment"]

results = []
for i in range(1, 11):
    metric = f"metric_{i}"
    c_vals = control[metric]
    t_vals = treatment[metric]
    
    # t-test
    t_stat, p_val = stats.ttest_ind(t_vals, c_vals, equal_var=False)
    
    results.append({
        "metric": metric,
        "control_mean": c_vals.mean(),
        "treatment_mean": t_vals.mean(),
        "p_value": p_val,
        "significant": p_val < 0.05
    })

results_df = pd.DataFrame(results)
print(f"Значимых метрик (p < 0.05): {results_df['significant'].sum()} из 10")
results_df
```

```python
alpha = 0.05
n_tests = 10
bonferroni_alpha = alpha / n_tests

print(f"Обычный порог: {alpha}")
print(f"Порог Бонферрони: {bonferroni_alpha}")

results_df["significant_bonferroni"] = results_df["p_value"] < bonferroni_alpha
print(f"Значимых метрик с поправкой: {results_df['significant_bonferroni'].sum()} из 10")
```
